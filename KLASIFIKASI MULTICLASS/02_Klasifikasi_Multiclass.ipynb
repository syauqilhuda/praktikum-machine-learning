## Tujuan
Notebook ini digunakan untuk mengidentifikasi tiga dataset yang termasuk klasifikasi **multiclass**. Klasifikasi multiclass adalah klasifikasi dengan satu kolom target yang memiliki lebih dari dua kategori kelas.

## Kriteria multiclass
Setiap baris data memiliki satu label target, tetapi pilihan label target berjumlah lebih dari dua kelas. Contohnya, target dengan label `1`, `2`, dan `3` termasuk multiclass.

````python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

pd.set_option("display.max_columns", 15)
pd.set_option("display.width", 140)

# Lokasi kerja notebook di VS Code
folder_kerja = Path.cwd()

# Mendukung notebook dijalankan dari folder proyek utama
# atau langsung dari folder KLASIFIKASI MULTICLASS
kandidat_folder = [
    folder_kerja / "dataset",
    folder_kerja / "KLASIFIKASI MULTICLASS" / "dataset"
]

folder_dataset = next(
    (folder for folder in kandidat_folder if folder.is_dir()),
    None
)

if folder_dataset is None:
    raise FileNotFoundError(
        "Folder dataset tidak ditemukan. Buat folder 'dataset' "
        "di dalam folder KLASIFIKASI MULTICLASS."
    )

print("Library berhasil diimpor.")
print("Folder kerja:", folder_kerja.resolve())
print("Folder dataset:", folder_dataset.resolve())
````

## 2. Fungsi Pemeriksaan Dataset

Fungsi ini menampilkan ukuran dataset, jumlah fitur, contoh data, nilai kosong, jumlah kelas target, distribusi kelas, dan grafik distribusi. Kolom identitas (jika ada) tidak dihitung sebagai fitur.


````python
def periksa_multiclass(df, kolom_target, nama_dataset, url_sumber,
                       kolom_id=None):
    print("=" * 75)
    print("NAMA DATASET:", nama_dataset)
    print("SUMBER:", url_sumber)
    print("=" * 75)

    if kolom_target not in df.columns:
        raise KeyError(
            f"Kolom target '{kolom_target}' tidak ditemukan. "
            f"Kolom tersedia: {list(df.columns)}"
        )

    data = df.dropna(subset=[kolom_target]).copy()
    y = data[kolom_target]

    kolom_dihapus = [kolom_target]
    if kolom_id and kolom_id in data.columns:
        kolom_dihapus.append(kolom_id)

    X = data.drop(columns=kolom_dihapus)

    print("\nUkuran data (baris, kolom):", df.shape)
    print("Jumlah data dengan target terisi:", len(data))
    print("Jumlah fitur yang digunakan:", X.shape[1])

    print("\nContoh data:")
    display(df.head())

    print("\nJumlah nilai kosong per kolom:")
    display(df.isnull().sum().to_frame("Jumlah Missing Value"))

    distribusi = y.value_counts().sort_index()
    print("\nJumlah kelas target:", y.nunique())
    print("Daftar kelas:", list(distribusi.index))
    print("\nDistribusi kelas:")
    display(distribusi.rename_axis("Kelas").reset_index(name="Jumlah Data"))

    plt.figure(figsize=(8, 4))
    distribusi.plot(kind="bar")
    plt.title(f"Distribusi Kelas - {nama_dataset}")
    plt.xlabel("Kelas")
    plt.ylabel("Jumlah Data")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()

    if y.nunique() > 2:
        print("KESIMPULAN: Dataset termasuk klasifikasi MULTICLASS.")
    else:
        print("PERHATIAN: Target dataset ini tidak memiliki lebih dari dua kelas.")

    return X, y
````

## 3. Dataset Pertama: Zoo

**Sumber:** UCI Machine Learning Repository – Zoo  
https://archive.ics.uci.edu/dataset/111/zoo

Dataset Zoo berisi karakteristik hewan, seperti memiliki rambut, bulu, telur, dan kemampuan hidup di air. Target `type` menunjukkan kelompok hewan. Dataset ini memiliki 101 data dan 7 kelas hewan, sehingga termasuk multiclass.


````python
file_zoo = folder_dataset / "zoo.data"

if not file_zoo.exists():
    raise FileNotFoundError(f"File tidak ditemukan: {file_zoo}")

kolom_zoo = [
    "animal_name", "hair", "feathers", "eggs", "milk", "airborne",
    "aquatic", "predator", "toothed", "backbone", "breathes",
    "venomous", "fins", "legs", "tail", "domestic", "catsize", "type"
]

df_zoo = pd.read_csv(file_zoo, header=None, names=kolom_zoo)

X_zoo, y_zoo = periksa_multiclass(
    df=df_zoo,
    kolom_target="type",
    nama_dataset="Zoo",
    url_sumber="https://archive.ics.uci.edu/dataset/111/zoo",
    kolom_id="animal_name"
)
````

### Alasan termasuk multiclass
Kolom target `type` mengelompokkan hewan ke dalam 7 jenis kelas. Setiap hewan memiliki satu label kelompok. Karena jumlah kelas target lebih dari dua, Zoo termasuk klasifikasi multiclass.


## 4. Dataset Kedua: Glass Identification

**Sumber:** UCI Machine Learning Repository – Glass Identification  
https://archive.ics.uci.edu/dataset/42/glass+identification

Dataset ini berisi hasil pengukuran komposisi kimia dan indeks bias sampel kaca. Target `Type` menunjukkan jenis kaca. Dataset UCI ini memiliki 214 data dan 6 kelas target yang teramati.


````python
file_glass = folder_dataset / "glass.data"

if not file_glass.exists():
    raise FileNotFoundError(f"File tidak ditemukan: {file_glass}")

kolom_glass = [
    "Id", "RI", "Na", "Mg", "Al", "Si",
    "K", "Ca", "Ba", "Fe", "Type"
]

df_glass = pd.read_csv(file_glass, header=None, names=kolom_glass)

X_glass, y_glass = periksa_multiclass(
    df=df_glass,
    kolom_target="Type",
    nama_dataset="Glass Identification",
    url_sumber="https://archive.ics.uci.edu/dataset/42/glass+identification",
    kolom_id="Id"
)
````

### Alasan termasuk multiclass
Kolom target `Type` menunjukkan kategori jenis kaca. Data yang tersedia memiliki enam kelas target yang teramati. Setiap sampel kaca hanya memiliki satu jenis label, sehingga dataset ini memenuhi kriteria klasifikasi multiclass.


## 5. Dataset Ketiga: Hayes-Roth

**Sumber:** UCI Machine Learning Repository – Hayes-Roth  
https://archive.ics.uci.edu/dataset/44/hayes+roth

Dataset Hayes-Roth berisi data kategorikal mengenai karakteristik subjek, seperti hobi, usia, tingkat pendidikan, dan status pernikahan. Target `class` memiliki tiga kategori, sehingga dataset ini termasuk multiclass.


````python
file_hayes = folder_dataset / "hayes-roth.data"

if not file_hayes.exists():
    raise FileNotFoundError(f"File tidak ditemukan: {file_hayes}")

kolom_hayes = [
    "name", "hobby", "age",
    "educational_level", "marital_status", "class"
]

df_hayes = pd.read_csv(
    file_hayes,
    header=None,
    names=kolom_hayes,
    skipinitialspace=True
)

X_hayes, y_hayes = periksa_multiclass(
    df=df_hayes,
    kolom_target="class",
    nama_dataset="Hayes-Roth",
    url_sumber="https://archive.ics.uci.edu/dataset/44/hayes+roth",
    kolom_id="name"
)
````

### Alasan termasuk multiclass
Kolom target `class` pada dataset Hayes-Roth memiliki tiga kategori kelas. Setiap baris data hanya memiliki satu kategori target. Oleh karena itu, dataset ini termasuk klasifikasi multiclass.


## 6. Kesimpulan

Berdasarkan pemeriksaan ketiga dataset:

1. **Zoo** termasuk multiclass karena target `type` memiliki 7 kelas hewan.
2. **Glass Identification** termasuk multiclass karena target `Type` memiliki 6 kelas jenis kaca yang teramati.
3. **Hayes-Roth** termasuk multiclass karena target `class` memiliki 3 kategori.

Ketiga dataset memiliki satu label target untuk setiap baris data dan jumlah kategori target lebih dari dua. Dengan demikian, ketiganya memenuhi karakteristik klasifikasi multiclass.
